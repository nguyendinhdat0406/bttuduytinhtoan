W6A17 
Nhập vào mảng vuông 2 chiều n hàng, n cột, in ra đường chéo chính và 
đường chéo phụ của mảng 

```python
n = int(input())
matrix = []
for i in range(n):
    matrix.append([])
    for j in range(n):
        matrix[i].append(input())
print("Các phần tử trên đường chéo chính:", end = "")
for i in range(n):
    print(matrix[i][i], end = " ")
print("Các phần tử trên đường chéo phụ:", end = "")
for i in range(n - 1, -1, -1):
    print(matrix[n-i-1][i], end = " ")
```

W6A18 
Nhập vào mảng n hàng, m cột số nguyên và số nguyên k (n, m, k nhập trên 
cùng một dòng, cách nhau dấu cách). Tính và in ra tổng các phần tử trên cột 
k (hàng k)

```python
n, m, k = map(int, input().split())
matrix = []
for i in range(n):
    matrix.append([])
    for j in range(m):
        matrix[i].append(float(input(f"Nhập phần tử hàng {i + 1} cột {j + 1}")))
print(matrix)
s1 = 0
s2 = 0
for i in range(n):
    s1 += matrix[i][k - 1]
print(f"Tổng các phần tử trên cột {k} là: {s1}")
for j in range(m):
    s2 += matrix[k - 1][j]
print(f"Tổng các phần tử trên hàng {k} là: {s2}")
```

W6A19 
Viết hàm đưa vào 1 list số nguyên, tìm và trả về vị trí có giá trị lớn nhất 
trong list

```python
def idxmax(ds):
    m = float('-inf')
    idx = 0
    for i in ds:
        if i > m:
            m = i
            idx = ds.index(i)
    return idx

print(idxmax([2, 4, 6, 1, 9, 5, 3]))
```

W6A20 
Viết hàm đưa vào một list số nguyên và một số nguyên dương k. Hãy tìm và 
trả về vị trí của phần tử đầu tiên có giá trị k trong list, nếu không có thì trả về -1 

```python
def idxk(ds, k):
    idx = -1
    for i in range(len(ds)):
        if ds[i] == k:
            idx = i
            break
    return idx

idxk([1, 5, 7, 8, 2, 5, 7], 5)
```

W6A21 
Nhập một dòng gồm các chuỗi (tách bởi dấu cách). Tạo dictionary ánh xạ 
chuỗi -> tuple các vị trí (0-based) mà chuỗi đó xuất hiện 

```python
ds = list(input().split())
print(ds)
d = {}
for i in ds:
    if i not in d:
        d.update({i : tuple([j for j in range(len(ds)) if ds[j] == i])})
print(d)
```

W6A24 
Viêết hàm đẩy tất cả các số 0 về đầu (về cuối), các số khác vẫn đúng vị trí 
trước sau: 0 2 3 2 4 0 0 0 9 => 0 0 0 2 3 2 4 9 

```python
def lendau(ds):
    ds1 = []
    ds2 = []
    for i in ds:
        if i == 0:
            ds1.append(0)
        else:
            ds2.append(i)
    return ds1 + ds2
print(lendau([0, 2, 3, 2, 4, 0, 0, 0, 9]))
```

W6A26 
Tính tổng các phần tử cực đại địa phương trong một danh sách, không tính 2 
phần tử đầu và cuối 

```python
def tongcucdai(ds):
    s = 0
    for i in range(1, len(ds)-1):
        if ds[i] > ds[i+1] and ds[i] > ds[i-1]:
            s += ds[i]
    return s
print(tongcucdai([1, 2, 1, 4, 5, 4, 7, 2]))
```

Cho một dãy số nguyên, Tìm và in ra vị trí các số < x (x là giá trị cho trước). 
nếu không có in ra -1 

```python
def index(ds, x):
    d = 0
    for i in range(len(ds)):
        if ds[i] < x:
            print(i, end = " ")
            d += 1
    if d == 0:
        return -1
    return
index([1, 2,3 ,4, 6, 7, 4, 2, 5], 4)
```

W6A28 
Cho một dãy các số nguyên, kiểm tra xem nó là dãy tăng thực sự hay giảm 
thực sự, hay vô danh (không tăng, không giảm) 

```python
def tanggiam(ds):
    if ds[0] < ds[1]:
        for i in range(1, len(ds) -1 ):
            if ds[i] >= ds[i+1]:
                return "Dãy vô danh"
        else:
            return "Dãy tăng"
    elif ds[0] > ds[1]:
        for i in range(1, len(ds) - 1):
            if ds[i] <= ds[i+1]:
                return "Dãy vô danh"
        else:
            return "Dãy giảm"
    else:
        return "Dãy vô danh"
tanggiam([5, 4, 3, 2, 1])
```
