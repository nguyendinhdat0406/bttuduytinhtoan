1 Yêu cầu. Cước phí Taxi Thông Minh. 
Viết một hàm duy nhất smart_taxi_fare(car_type, start_hour, km, wait_minutes) tính và trả về total (tổng tiền phải trả, 
kiểu số nguyên, đơn vị VND). 
 
Các quy tắc cần cài đặt: 
1. Giá mở cửa (0–0.8 km): - Xe 4 chỗ: 11,000 VND - Xe 7 chỗ: 12,000 VND 
2. Giá theo km sau 0.8 km: - Xe 4 chỗ: từ 0.8→20 km nh 13,000 VND/km; trên 20 km tính 10,500 VND/km - Xe 7 chỗ: từ 0.8→20 km nh 14,000 VND/km; trên 20 km tính 12,000 VND/km 
3. Phí chờ: 5 phút đầu miễn phí; sau đó, mỗi 5 phút (làm tròn lên theo từng block 5 phút) tính 2,000 VND. 
4. Phụ thu ban đêm: nếu start_hour thuộc [22,23,0,1,2,3,4,5], cộng +20% trên tiền tạm tính (mở cửa + quãng đường + 
chờ). 
5. Điều kiện đặc biệt: - Nếu km > 30 và không phải ban đêm, áp dụng giảm 10% trên tổng tiền. - Nếu tổng tiền cuối cùng < 50,000 nhưng km > 5, đặt tổng tiền bằng 50,000 VND (mức tối thiểu).

```python
import math
def smart_taxi_fare(car_type, start_hour, km, wait_minutes):
    total = 0
    # Xe 4 chỗ
    if car_type == 4:
        # Xử lý giá xe tính theo quãng đường
        if 0 < km <= 0.8:
            total += 11000
        elif 0.8 < km <= 20:
            total += 11000 + (km - 0.8) * 13000
        elif km > 20:
            total += 11000 + 19.2 * 13000 + (km - 20) * 10500
    
    # Xe 7 chỗ
    elif car_type == 7:
        if 0 < km <= 0.8:
            total += 12000
        elif 0.8 < km <= 20:
            total += 12000 + (km - 0.8) * 14000
        elif km > 20:
            total += 12000 + 19.2 * 14000 + (km - 20) * 12000
    
    # Thời gian chờ
    if wait_minutes > 5:
        total += math.ceil((wait_minutes - 5)/5) * 2000
    # Phụ thu ban đêm 
    if start_hour in [22, 23, 0, 1, 2, 3, 4, 5]:
        total = total * 1.2
    # Điều kiện đặc biệt
    if km > 30 and (start_hour in [22, 23, 0, 1, 2, 3, 4, 5]) == False:
        total = total * 0.9
    if total < 50000 and km > 5:
        total = 50000
    return int(total)

smart_taxi_fare(7, 23, 32, 17)
```

dictionary) 

```python
def max_min(a, b, c):
    ds = [a, b, c]
    mx = a
    mn = a
    for i in ds:
        if mx < i:
            mx = i
        if mn > i:
            mn = i
    return [mx, mn]

max_min(2,5,1)
```

16 Viết hàm trả về góc ∠ABC theo đơn vị độ khi biết tọa độ các điểm A(x1,y1), B(x2,y2), C(x3,y3). 

```python
import math
def goc(x1, y1, x2, y2, x3, y3):
    v1 = [x1 - x2, y1 - y2]
    v2 = [x3 - x2, y3 - y2]
    cos = abs((v1[0]*v2[0] + v1[1]*v2[1]) / ((v1[0]**2 + v1[1]**2)**0.5 * (v2[0]**2 + v2[1]**2)**0.5))
    return math.acos(cos)*180/math.pi

goc(2,4,5,1,2,1)
```

25 Yêu cầu: Viết hàm point_in_quadrilateral(quad, p) trả về True nếu điểm p nằm bên trong hoặc trên biên của tứ giác quad, 
và False nếu ngược lại. - Đầu vào: - quad: danh sách hoặc bộ gồm đúng 4 đỉnh [(x1,y1), (x2,y2), (x3,y3), (x4,y4)] theo thứ tự (tứ giác có thể lồi hoặc lõm, 
nhưng không tự cắt). - p: một điểm (x, y). - Đầu ra: giá trị boolean. - Gợi ý: Cài đặt phép kiểm tra điểm trong đa giác tổng quát (ví dụ: ray casting), đồng thời tính cả trường hợp điểm nằm 
trên biên là ở trong. 

```python
def point_in_quadrilateral(quad, p):
    v1 = [p[0] - quad[0][0], p[1] - quad[0][1]]
    v2 = [p[0] - quad[1][0], p[1] - quad[1][1]]
    v3 = [p[0] - quad[2][0], p[1] - quad[2][1]]
    v4 = [p[0] - quad[3][0], p[1] - quad[3][1]]
    vecto1 = [quad[1][0] - quad[0][0], quad[1][1] - quad[0][1]]
    vecto2 = [quad[3][0] - quad[0][0], quad[3][1] - quad[0][1]]
    
    tugiac = abs(vecto1[0]*vecto2[1] - vecto2[0]*vecto1[1])
    tamgiac1 = 1/2 * abs(v1[0]*v2[1] - v2[0]*v1[1])
    tamgiac2 = 1/2 * abs(v2[0]*v3[1] - v3[0]*v2[1])
    tamgiac3 = 1/2 * abs(v3[0]*v4[1] - v4[0]*v3[1])
    tamgiac4 = 1/2 * abs(v1[0]*v4[1] - v4[0]*v1[1])

    if tugiac >= (tamgiac1 + tamgiac2 + tamgiac3 + tamgiac4):
        return True
    else:
        return False
    
point_in_quadrilateral([(0,0), (4,0), (4,3), (0,3)], [2, 1])
```

26 Yêu cầu: Viết hàm are_collinear(x1, y1, x2, y2, x3, y3) trả về True nếu ba điểm thẳng hàng, ngược lại trả về False. - Đầu vào: ba điểm x1, y1, x2, y2, x3, y3 biểu diễn tọa độ của ba điểm. - Đầu ra: giá trị boolean. 
 
- Gợi ý: Dùng phép kiểm tra diện tích (tích có hướng) và một sai số nhỏ khi so sánh số thực. Các điểm nằm trên cùng một 
đường thẳng được xem là thẳng hàng (kể cả khi có điểm trùng nhau). 

```python
def are_collinear(x1,y1,x2,y2,x3,y3):
    v1 = (x2 - x1, y2 - y1)
    v2 = (x3 - x1, y3 - y1)
    if v1[0] * v2[1] - v2[0] * v1[1] != 0:
        return False
    else:
        return True
    
are_collinear(0,1,2,0,1,0.5)
```

27 Yêu cầu: Viết hàm quadratic_roots_count(a, b, c) trả về số nghiệm thực của phương trình ax^2 + bx + c = 0. - Đầu vào: các số thực a, b, c. - Đầu ra: 0, 1, hoặc 2. Nếu phương trình suy biến thành 0 = 0 (tức a == 0, b == 0, c == 0), trả về float('inf'). - Quy tắc: - Nếu a != 0, dùng biệt thức D = b^2 - 4ac. - Nếu a == 0 và b != 0, đó là phương trình bậc nhất với đúng một nghiệm thực. - Nếu a == 0 và b == 0 và c != 0, phương trình vô nghiệm. 

```python
def quadratic_roots_count(a, b, c):
    if a == 0 and b == 0 and c == 0:
        return float("inf")
    else:
        if a == 0:
            if b != 0:
                return "1"
            elif b == 0 and c != 0:
                return "0"
        else:
            if b**2 - 4*a*c < 0:
                return "0"
            elif b**2 - 4*a*c == 0:
                return "1"
            else:
                return "2"
            
quadratic_roots_count(1, 2, -5)
```

```python
def computer(a, b, op):
    if op == "/" and b == 0:
        return ValueError
    elif (op in ["+", "-", "*", "/"]) == False:
        return ValueError
    else:
        if op == "+":
            print(a + b)
        elif op == "-":
            print(a - b)
        elif op == "*":
            print(a * b)
        else:
            print(a / b)
    
computer(6, 3, "+")
```

35 Yêu cầu: Viết hàm divide_with_remainder(dividend, divisor) trả về bộ (q, r) trong đó q là thương nguyên và r là số dư sao 
cho dividend = divisor*q + r với 0 <= |r| < |divisor| (dùng ngữ nghĩa của Python với // và %). - Đầu vào: số nguyên dividend, divisor (divisor khác 0). - Đầu ra: bộ hai số nguyên (q, r). - Ví dụ: divide_with_remainder(5, 3) -> (1, 2). 

```python
def divide_with_remaindef(dividend, divisor):
    for i in range(divisor):
        q = (dividend - i)/divisor
        if q % 1 == 0:
            print(((dividend - i)/divisor, i), end= ",")
    return

divide_with_remaindef(5, 3)
```

41 Viết hàm trả về ký tự đứng sau ký tự x n vị trí trong bảng chữ cái. 
Giữ nguyên chữ hoa/chữ thường và quay vòng trong phạm vi A..Z hoặc a..z. 

```python
def chucai(x, n):
    if ("a" <= x <= "z" and "a" <= chr(ord(x) + n) <= "z") or ("A" <= x <= "Z" and "A" <= chr(ord(x) + n) <= "Z"):
        return chr(ord(x) + n)
    else:
        if "a" <= x <= "z":
            return chr(ord("a") + n - (ord("z") - ord(x)) - 1)
        elif "A" <= x <= "Z":
            return chr(ord("A") + n - (ord("Z") - ord(x)) - 1)
        
chucai("A",26)
```

```python
def giaithua(n):
    if n == 0:
        return 1
    else:
        p = 1
        for i in range(1, n + 1):
            p = p * i
        return p

giaithua(4)
```
