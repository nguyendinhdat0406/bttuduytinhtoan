Xây dựng lớp Shape với phương thức area(). 
Xây dựng lớp Rectangle(w, h) và Circle(r) kế thừa từ Shape, ghi đè area(). 
Lấy π = 3.14. 
 
Input 
Dòng 1: loại hình (RECTANGLE hoặc CIRCLE) 
Dòng 2: tham số tương ứng. 
 
Output 
In ra diện tích với 2 chữ số thập phân.

```python
class Shape():
    def area(self):
        pass

class Retangle(Shape):
    def __init__(self, w, h):
        self.w = w
        self.h = h
    def area(self):
        return self.w * self.h
    
class Circle(Shape):
    def __init__(self, r):
        self.r = r
    def area(self):
        return 3.14 * self.r**2
    
h1 = Retangle(3, 4)
print(h1.area())
t1 = Circle(3)
print(t1.area())
```

Dòng 1: loại đối tượng (Vehicle hoặc Car) 
Dòng 2: dữ liệu tương ứng. 
 
Output 
In ra chuỗi info().

```python
class Vehicle():
    def __init__(self, brand, speed):
        self.brand = brand 
        self.speed = speed 
    def info(self):
        return f"{self.brand} - {self.speed}"

class Car(Vehicle):
    def __init__(self, brand, speed, seats):
        super().__init__(brand, speed)
        self.seats = seats 
    def info(self):
        return f"{self.brand} - {self.speed} - {self.seats}"

car1 = Car("matda", 200, "bsw")
print(car1.info())
```

W8A20 Yêu cầu 
Xây dựng lớp Account(balance) với hai phương thức deposit(x) và get_balance(). 
Xây dựng lớp SavingAccount(balance, rate) kế thừa từ Account, thêm phương thức 
add_interest() để cộng balance*rate vào số dư. 
 
Input 
balance rate 
 
Output 
Số dư sau khi cộng lãi, in 2 chữ số thập phân. 

```python
class Account():
    def __init__(self, balance):
        self.balance = balance
    def deposit(self, x):
        self.balance += x
        return "Đã gửi tiền"
    def get_balance(self):
        return f"Số dư tài khoản: {self.balance}"

class SavingAccount(Account):
    def __init__(self, balance, rate):
        super().__init__(balance)
        self.rate = rate
    def add_interest(self):
        self.balance = self.balance * (1 + self.rate)
        return self.balance

no1 = SavingAccount(10000000, 0.05)
print(no1.deposit(5000000))
print(no1.get_balance())
print(no1.add_interest())
```

W8A22 Yêu cầu 
Xây dựng lớp Matrix2x2(a, b, c, d) biểu diễn ma trận: 
[a b] 
[c d] 
Viết phương thức trace() trả về a + d và determinant() trả về ad - bc. 
 
Input 
Một dòng chứa 4 số nguyên a b c d. 
 
Output 
In ra trace và determinant trên cùng một dòng.

```python
class Matrix2x2():
    def __init__(self, a, b, c, d):
        self.a = a
        self.b = b
        self.c = c
        self.d = d
    def trace(self):
        return self.a + self.d
    def determinant(self):
        return self.a * self.d - self.b * self.c

a, b, c, d = map(int, input().split())
no1 = Matrix2x2(a, b, c, d)
print(no1.trace())
print(no1.determinant())
```

W8A27 Yêu cầu 
Xây dựng lớp Product(name, price). 
Xây dựng lớp DiscountProduct(name, price, percent) kế thừa từ Product, ghi đè phương thức 
final_price() để trả về giá sau giảm. 
Công thức: price * (100 - percent) / 100 
 
Input 
name price percent 
 
Output 
In ra giá cuối cùng với 2 chữ số thập phân. 

```python
class Product():
    def __init__(self, name, price):
        self.name = name
        self.price = price
    def final_price(self):
        return self.price

class DiscountProduct(Product):
    def __init__(self, name, price, percent):
        super().__init__(name, price)
        self.percent = percent
    def final_price(self):
        return self.price * (100 - self.percent) / 100

no1 = DiscountProduct("a", 100000, 10)
print(no1.final_price())
```

W8A28 Yêu cầu 
Xây dựng lớp Counter có thuộc tính count ban đầu bằng 0 và các phương thức inc(), dec(), 
value(). 
Sau đó thực hiện dãy thao tác trên đối tượng. 
 
Input 
Dòng 1: n 
Dòng 2: n thao tác, mỗi thao tác là INC hoặc DEC, cách nhau bởi dấu cách. 
 
Output 
Giá trị cuối cùng của count.

```python
class Counter():
    def __init__(self, count = 0):
        self.count = 0
    def inc(self):
        self.count += 1
        return "Đã đếm"
    def dec(self):
        self.count -= 1
        return "Đã đếm"
    def value(self):
        return self.count

n = int(input())
lst = list(input().split())
my_counter = Counter()
for i in lst:
    if i == 'inc':
        my_counter.inc()
    elif i == 'dec':
        my_counter.dec()
print(my_counter.value())
```

W8A29 Yêu cầu 
Xây dựng lớp Time(h, m, s) với phương thức to_seconds() đổi thời gian sang tổng số giây. 
 
Input 
h m s 
 
Output 
Tổng số giây.

```python
class Time():
    def __init__(self, h, m, s):
        self.h = h
        self.m = m
        self.s = s
    def to_seconds(self):
        return self.h * 3600 + self.m * 60 + self.s

h, m, s = map(int, input().split())
no1 = Time(h, m, s)
print(no1.to_seconds())
```

W8A30 Yêu cầu 
Xây dựng lớp Stack sử dụng danh sách để cài đặt ngăn xếp với các phương thức push(x), 
pop(), top(). 
Dữ liệu đảm bảo mọi lệnh pop và top đều hợp lệ. 
 
Input 
Dòng 1: n 
n dòng tiếp theo là một trong ba lệnh: 
PUSH x 
POP 
TOP 
 
Output 
Mỗi lệnh TOP in ra giá trị ở đỉnh ngăn xếp trên một dòng. 

```python
class Stack():
    def __init__(self, lst = []):
        self.lst = lst
    def push(self, x):
        self.lst.append(x)
    def pop(self):
        self.lst.pop()
    def top(self):
        return self.lst[-1]
    def show(self):
        return self.lst

n = int(input())
k = ""
my_stack = Stack()
for i in range(n):
    k = input()
    if k == "push":
        x = input("Nhập x: ")
        my_stack.push(x)
    elif k == "pop":
        my_stack.pop()
    elif k == "top":
        print(my_stack.top())
print(my_stack.show())
```
