Cho một mảng số nguyên không âm. Gọi mảng đó là "mảng đẹp" nếu 
mỗi phần tử x trong mảng xuất hiện đúng x lần. 
Kiểm tra xem mảng có phải là mảng đẹp hay không. Nếu không hãy xóa 
đi y số (y là nhỏ nhất có thể) để mảng còn lại thành mảng đẹp.

```python
lst = list(map(int, input().split()))
test = True
for i in set(lst):
    if i != lst.count(i):
        test = False
        break
else:
    print("Mảng đẹp")

if test == False:
    for i in set(lst) and i == 0:
        if i > lst.count(i):
            lst = [x for x in lst if x != i]
        elif i < lst.count(i):
            while i != lst.count(i):
                lst.remove(i)
    print(lst)
```

W7A17 
Tìm phần tử xuất hiện nhiều hơn n/2 lần trong mảng (nếu có) với độ 
phức tạp thời gian là O(n), độ phức tạp bộ nhớ là O(1), - Boyer–Moore. 
Nếu không có phần tử nào thỏa mãn thì không in gì 

```python
lst = list(map(str, input().split()))
l = list(set(lst))
dem = [0] * len(l)
for i in lst:
    dem[l.index(i)] += 1
for i in range(len(dem)):
    if dem[i] > len(lst)/2:
        print(l[i])
```

W7A22 
Cho N tên sinh viên gồm cả họ và tên, tìm xem tên nào hay gặp nhất. 
Nếu có nhiều tên cùng số lần gặp thì lấy tên có thứ tự từ điển nhỏ nhất 

```python
n = int(input())
name = []
for i in range(n):
    name.append(input().strip())
name.sort()
dic = {}
for i in list(set(name)):
    dic.update({i : name.count(i)})
dem = 0
for i in dic:
    if dic[i] > dem:
        dem = dic[i]
print(dem)
```

Tìm số có đúng k chữ số lớn nhất có thể trong một số có n chữ số (coi 
đầu vào là một số rất lớn - str). Không được hoán đổi vị trí các chữ số

```python
n = input()
k = int(input())
smax = 0
num = 0
for i in range(len(n) - k + 1):
    s = 0
    for j in range(i, i + k):
        s += int(n[j])
    if s > smax:
        smax = s
        num = int(n[i: i + k])
print(num)
```

W7A27 
Cho một dãy ngoặc ()(){}{}[][]..., kiểm tra xem dãy ngoặc đã cho có hợp lệ 
hay không 

```python
s = input()
test = True
if len(s) % 2 != 0:
    print("Không hợp lệ")
else:
    for i in range(len(s)):
        if s[i] == "(":
            if ")" not in s:
                test = False
                print("Không hợp lệ")
            else:
                for j in range(i + 1, len(s)):
                    if s[j] == ")" and (j - i) % 2 == 0:
                        test = False
                        print("Không hợp lệ")
                        break
        elif s[i] == "{":
            if "}" not in s:
                test = False
                print("Không hợp lệ")
            else:
                for j in range(i + 1, len(s)):
                    if s[j] == "}" and (j - i) % 2 == 0:
                        test = False
                        print("Không hợp lệ")
                        break
        elif s[i] == "[":
            if "]" not in s:
                test = False
                print("Không hợp lệ")
            else:
                for j in range(i + 1, len(s)):
                    if s[j] == "]" and (j - i) % 2 == 0:
                        test = False
                        print("Không hợp lệ")
                        break
        if test == False:
            break
    else:
        print("Hợp lệ")
```

```python
s = input()
dic = {")": "(", "}" : "{", "]": "["}
stack = []
for i in s:
    if i in ["(", "{", "["]:
        stack.append(i)
    else:
        if dic[i] == stack[-1]:
            stack.pop(-1)
        else:
            print("Không hợp lệ")
            break
else:
    if len(stack) == 0:
        print("Hợp lệ")
    else:
        print("Không hợp lệ")
#([{}])
```

```python
numbers = input().strip()
n = int(input())
def songuyento(number):
    for i in range(2, number):
        if number % i == 0:
            return True
    else:
        return False
hopso = ""
s = ""
d = 0
for i in numbers:
    if int(i) > n and i != 1 and songuyento(int(i)) == True:
        s += i
    else:
        if len(s) > d:
            d = len(s)
            hopso = s
            s = ""
print(hopso)

#132456697
#n = 3
```

W7A30 
Tìm Dãy con tăng dần KHÔNG liên tục dài nhất của một danh sách nhập vào từ bàn phím  

```python
lst = list(map(int, input().split()))
result = []

for i in range(len(lst)):
    x = [lst[i]]
    for j in range(i + 1, len(lst)):
        if lst[j] > x[-1]:
            x.append(lst[j])
        elif lst[j] < x[-1] and len(x) >= 2:
            if lst[j] > x[-2]:
                x[-1] = lst[j]
    if len(x) > len(result):
        result = x

print(result)
# input: 10 22 9 33 21 50 41 60 80
```
