Cho một mảng số nguyên không âm. Gọi 
mảng đó là "mảng đẹp" nếu mỗi phần tử 
x trong mảng xuất hiện đúng x lần. 
Kiểm tra xem mảng có phải là mảng đẹp 
hay không. Nếu không hãy xóa đi y số (y 
là nhỏ nhất có thể) để mảng còn lại thành 
mảng đẹp. 

```python
arr = list(map(int, input().split()))
dem = 0
for i in set(arr):
    d = 0
    for j in arr:
        if i == j:
            d += 1
    if d != i:
        if d > i:
            dem += d - i
        else:
            dem += d
if d == 0:
    print("Yes")
else:
    print("No", dem)
```

Tìm phần tử xuất hiện nhiều hơn n/2 lần 
trong mảng (nếu có) với độ phức tạp thời 
gian là O(n), độ phức tạp bộ nhớ là O(1), - Boyer–Moore. Nếu không có phần tử 
nào thỏa mãn thì không in gì 

```python
lst = list(map(int, input().split()))
num = lst[0]
d = 0
for i in lst:
    if i == num:
        d += 1
    else:
        if d != 0:
            d -= 1
        else:
            num = i
if lst.count(num) > len(lst)/2:
    print(num)
```

Sắp xếp có điều kiện: Số chẵn đứng 
trước, số lẻ đứng sau, chẵn tăng dần, lẻ 
giảm dần

Cho N tên sinh viên gồm cả họ và tên, 
tìm xem tên nào hay gặp nhất. Nếu có 
nhiều tên cùng số lần gặp thì lấy tên có 
thứ tự từ điển nhỏ nhất

```python
n = int(input())
names = []
for i in range(n):
    name = list(input().split())
    names.append(name[-1])
names.sort()
d = 0
max_name = ""
for i in set(names):
    if names.count(i) > d:
        max_name = i
        d = names.count(i)
print(max_name)
```

W7A25 
Tìm số có đúng k chữ số lớn nhất có thể 
trong một số có n chữ số (coi đầu vào là 
một số rất lớn - str). Không được hoán 
đổi vị trí các chữ số 

```python
num = input().split()
k = int(input())
number = ""
idx = -1
while len(number) != k:
    n = 0
    for i in range(idx + 1, len(num) - (k - len(number)) + 1):
        if int(num[i]) > int(n):
            n = num[i]
            idx = i
    number += n

print(int(number))
```

Tách các số có trong một đoạn văn bản, 
tính tổng và tính trung bình các số đó.  

```python
s = input().strip()
S = 0
d = 0
i = 0
while i < len(s):
    if s[i].isdigit() == True:
        j = 0
        while i + j < len(s) and s[i + j].isdigit() == True:
            j += 1
        S += int(s[i : i + j])
        d += 1
        i += j
    else:
        i += 1

print(S)
print(S/d)
```

W7A27 
Cho một dãy ngoặc ()(){}{}[][]..., kiểm tra 
xem dãy ngoặc đã cho có hợp lệ hay 
không 

```python
s = input().strip()
stack = []
dic = {")" : "(", "}" : "{", "]" : "["}
for i in s:
    if i in ["(", "[", "{"]:
        stack.append(i)
    elif i in [")", "]", "}"]:
        if len(stack) != 0:
            if dic[i] == stack[-1]:
                stack.pop(-1)
        else:
            stack.append(1)
            break

if len(stack) == 0:
    print("Yes")
else:
    print("No")
```

Bà ̀ i toán Tìm dãy các hợp số dài nhất liên 
tiếp không chứa số nguyên nào <= N (N 
là số nguyên dương cho trước). Nếu có 
nhiều dãy cùng độ dài thì lấy dãy nhỏ 
nhất (gặp đầu tiên từ trái qua)

```python
n = int(input())
lst = []
for i in range(4, n):
    for j in range(2, i):
        if i % j == 0:
            lst.append(i)
            break

nums = []
arr = []
for i in lst:
    if len(arr) == 0:
        arr.append(i)
    else:
        if arr[-1] + 1 == i:
            arr.append(i)
        else:
            if len(arr) > len(nums):
                nums = arr
            arr = [i]
    if len(arr) > len(nums):
        nums = arr
print(nums)
```

Tìm Dãy con tăng dần liên tục dài nhất 
của một danh sách nhập vào từ bàn 
phím. Lấy dãy đầu tiên nếu có nhiều dãy 
có số phần tử bằng nhau 

```python
numbers = list(map(int, input().split()))
max_leng = 0
max_start = 0
leng = 1
start = 0
for i in range(1, len(numbers)):
    if numbers[i] > numbers[i - 1]:
        leng += 1
    else:
        if leng > max_leng:
            max_leng = leng
            max_start = start 
        leng = 1
        start = i
    if leng > max_leng:
            max_leng = leng
            max_start = start 
print(numbers[max_start : max_start + max_leng])
# 1 2 3 4 5 6 1 2 3 4 9 10 12
```

W7A30 
Tìm Dãy con tăng dần KHÔNG liên tục 
dài nhất của một danh sách nhập vào từ 
bàn phím 

```python
lst = list(map(int, input().split()))
result = []
for i in range(len(lst)):
    arr = [lst[i]]
    for j in range(i + 1, len(lst)):
        if lst[j] >= arr[-1]:
            arr.append(lst[j])
    if len(arr) > len(result):
        result = arr
print(result)
```
