Bài 5: Viết chương trình giải phương trình bậc 2 (ax2 + bx + c = 0)

```python
a, b, c = map(float, input().split())
denta = b**2 - 4*a*c
if a == 0:
    print("Nghiệm:", -c/b)
else:
    if denta > 0:
        x1 = (-b + (denta)**0.5)/(2*a)
        x2 = (-b - (denta)**0.5)/(2*a)
        print(f"{x1} \n {x2}")
    elif denta == 0:
        x = -b/(2*a)
        print(x)
    else:
        print("Vô nghiệm")
```

Bài 8: Giải hệ phương trình tuyến tính: 
ax + by = m 
cx + dy = n

```python
a,b,m = map(float, input("Nhập các hệ số của phương trình thứ nhất: ").split())
c, d, n = map(float, input("Nhập các hệ số của phương trình thứ hai: ").split())
if a == c and b == d and m == n:
    print("Phương trình có vô số nghiệm")
elif a == c and b == d:
    print("Phương trình vô nghiệm ")
else:
    x = 0
    y = 0
    l1 = [a, b, m]
    l2 = [c, d, n]
    l3 = []
    for i in range(3):
        l3.append(a*l2[i] - c*l1[i])

    y += l3[2] / l3[1]
    x += (m - b * y)/a
    print( f"Nghiệm của hệ phương trình là x = {x}, y = {y}")
```

Bài 9: Nhập vào thời gian 1 công việc nào đó là x giây. Hãy chuyển đổi và viết 
ra màn hình số thời gian trên dưới dạng bằng bao nhiêu giờ, bao nhiêu phút, bao 
nhiêu giây.  

```python
x = float(input())
m = x // 60
hour = m // 60
second = x - m*60
min = m - hour*60
print(f"{hour} giờ {min} phút {second} giây")
```
